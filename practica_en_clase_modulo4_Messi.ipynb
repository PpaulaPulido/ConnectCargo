```python
import pandas as pd
import zipfile
import io
from google.colab import files

# 1. Subir el archivo (Esto ya lo tienes)
uploaded = files.upload()

# Obtener el nombre del archivo ZIP subido
nombre_zip = list(uploaded.keys())[0]
print(f"\n Procesando archivo comprimido: {nombre_zip}")

# 2. Abrir el ZIP y buscar el CSV correcto
with zipfile.ZipFile(nombre_zip, 'r') as z:
    # Listar qué archivos hay adentro
    archivos_internos = z.namelist()
    print(f" Archivos encontrados dentro del ZIP: {archivos_internos}")

    # Lógica para elegir: Buscamos el que diga "v2" (versión 2)
    # Si no hay v2, tomamos el primero que sea .csv
    archivo_a_leer = next((f for f in archivos_internos if 'v2' in f), None)

    if not archivo_a_leer:
        # Si no encuentra 'v2', toma el primer csv que vea
        archivo_a_leer = [f for f in archivos_internos if f.endswith('.csv')][0]

    print(f" Seleccionado para lectura: {archivo_a_leer}")

    # 3. Leer el archivo específico
    with z.open(archivo_a_leer) as f:
        df = pd.read_csv(f, low_memory=False) # low_memory=False ayuda si el archivo es grande y sucio

print(f"\n ¡Listo! Dataset cargado.")
print(f" Filas: {df.shape[0]}, Columnas: {df.shape[1]}")
display(df.head(3))
```

**¿Por qué existe columnas numericas que me toma como texto?**

*    Python ve el símbolo de euro (€) o las letras (lbs, K) y decide que toda la columna es texto.



```python
# BLOQUE DE LIMPIEZA 1: DINERO
# Objetivo: Convertir texto como "€1.5M" a numero 1500000.0

def conversion_dinero(valor):
    if not isinstance(valor, str):
        return valor

    # Limpiamos simbolo de moneda y espacios
    valor_limpio = valor.replace('€', '').strip()

    multiplicador = 1
    if 'M' in valor_limpio:
        multiplicador = 1000000
        valor_limpio = valor_limpio.replace('M', '')
    elif 'K' in valor_limpio:
        multiplicador = 1000
        valor_limpio = valor_limpio.replace('K', '')

    try:
        return float(valor_limpio) * multiplicador
    except ValueError:
        return 0.0

# Aplicamos la funcion a las 3 columnas financieras
columnas_dinero = ['Value', 'Wage', 'Release Clause']

for col in columnas_dinero:
    df[col] = df[col].apply(conversion_dinero)

print(" Dinero convertido a numeros. Ejemplo:")
print(df[['Name', 'Value', 'Wage']].head(3))
```

```python
# BLOQUE DE LIMPIEZA 2: FISICO
# Objetivo: Unificar Altura en cm y Peso en kg

def limpiar_altura(x):
    if isinstance(x, str):
        if 'cm' in x:
            return int(x.replace('cm', ''))
        elif "'" in x: # Si esta en pies y pulgadas (ej: 5'7")
            pies, pulgadas = x.split("'")
            pulgadas = pulgadas.replace('"', '')
            return round(float(pies) * 30.48 + float(pulgadas) * 2.54)
    return x

def limpiar_peso(x):
    if isinstance(x, str):
        if 'kg' in x:
            return int(x.replace('kg', ''))
        elif 'lbs' in x: # Si esta en libras
            return round(float(x.replace('lbs', '')) * 0.453592)
    return x

df['Height'] = df['Height'].apply(limpiar_altura)
df['Weight'] = df['Weight'].apply(limpiar_peso)

print(" Altura y Peso limpios. Ejemplo:")
print(df[['Name', 'Height', 'Weight']].head(3))
```

```python
# BLOQUE DE LIMPIEZA 3: ESTRELLAS Y HITS
# Objetivo: Quitar caracteres especiales de habilidades y trafico

# 1. Limpiar Estrellas
cols_estrellas = ['W/F', 'SM', 'IR']
for col in cols_estrellas:
    # Usamos .astype(str) para asegurar que sea texto antes de reemplazar
    df[col] = df[col].astype(str).str.replace('★', '').str.strip()
    # Convertimos a numero
    df[col] = pd.to_numeric(df[col], errors='coerce').fillna(0).astype(int)

# 2. Limpiar Hits (Vistas)
def limpiar_hits(x):
    if isinstance(x, str):
        x = x.lower()
        if 'k' in x:
            return int(float(x.replace('k', '')) * 1000)
        return int(float(x))
    return x

df['Hits'] = df['Hits'].apply(limpiar_hits).fillna(0).astype(int)

print(" Habilidades y Hits limpios. Ejemplo:")
print(df[['Name', 'W/F', 'Hits']].head(3))
```

```python
# VERIFICACION FINAL
print("--- ¿Donde estan ahora nuestras columnas problematicas? ---")
nuevas_numericas = df.select_dtypes(include=['number']).columns.tolist()

# Verificamos si 'Value' ya esta en la lista de numeros
if 'Value' in nuevas_numericas:
    print(" EXITO: 'Value' ahora es NUMERICO.")
else:
    print("ERROR: 'Value' sigue siendo texto.")

if 'Height' in nuevas_numericas:
    print(" EXITO: 'Height' ahora es NUMERICO.")

print("\nEstadisticas reales (ahora si funcionan):")
print(df[['Value', 'Wage', 'Height', 'Weight']].describe())
```

```python
# BLOQUE  LIMPIEZA: FECHAS Y ESTADISTICAS
# Objetivo: Convertir 'Joined' a fecha y ver los promedios reales de dinero

# 1. Convertir la columna 'Joined' a formato Fecha (Datetime)
df['Joined'] = pd.to_datetime(df['Joined'])

print(" Columna 'Joined' convertida a Fecha.")
print(df['Joined'].head(3))

# 2. Creacion de una variable nueva (Ingenieria de Caracteristicas basica)
# Calculamos cuantos años lleva el jugador en el club (asumiendo fecha actual 2021 del dataset)
df['Years_Active'] = 2026 - df['Joined'].dt.year

# 3. ¡LA PRUEBA DE FUEGO! -> Estadísticas del Dinero
# Ahora que 'Value' y 'Wage' son numeros, .describe() nos dara promedios y maximos
pd.options.display.float_format = '{:,.0f}'.format # Formato para leer millones sin notacion cientifica

print("\n--- ESTADISTICAS FINALES DEL DATASET LIMPIO ---")
columnas_interes = ['Value', 'Wage', 'Release Clause', 'Height', 'Weight', 'Years_Active']
print(df[columnas_interes].describe())
```

```python
# ESTRATEGIA CORREGIDA PARA SALVAR CLUB Y NACIONALIDAD

columnas_one_hot = []
columnas_frequency = []
columnas_eliminar = []

print("\n  Re-evaluando estrategia de texto...")

for columna in columnas_texto:
    num_categorias = df[columna].nunique()

    # Mantenemos One-Hot para pocas categorías (ideal para arboles de decisión)
    if num_categorias <= 15:
        columnas_one_hot.append(columna)
        print(f"    [One-Hot]   {columna}: {num_categorias} cats")

    # CAMBIO CLAVE: Subimos el límite a 1000 para incluir Club y Nacionalidad
    # Esto salvará Club (682) y Nationality (164), pero seguirá borrando Nombres (18k)
    elif num_categorias <= 1000:
        columnas_frequency.append(columna)
        print(f"    [Frequency] {columna}: {num_categorias} cats (SALVADA)")

    else:
        columnas_eliminar.append(columna)
        print(f"    [Eliminar]  {columna}: {num_categorias} cats (Ruido/ID)")

print(f"\n Estrategia MEJORADA:")
print(f"   One-Hot Encoding: {columnas_one_hot}")
print(f"   Frequency Encoding: {columnas_frequency}")
print(f"   Eliminar: {columnas_eliminar}")
```

```python
# Analizar columnas de texto para decidir estrategias
columnas_one_hot = []
columnas_frequency = []
columnas_eliminar = []

print("\n Analizando columnas de texto...")
for columna in columnas_texto:
    num_categorias = df[columna].nunique()

    if num_categorias <= 15:
        columnas_one_hot.append(columna)
        print(f"    {columna}: {num_categorias} categorías → One-Hot Encoding")
    elif num_categorias <= 1000:
        columnas_frequency.append(columna)
        print(f"    {columna}: {num_categorias} categorías → Frequency Encoding")
    else:
        columnas_eliminar.append(columna)
        print(f"    {columna}: {num_categorias} categorías → Demasiadas, se eliminará")

print(f"\n Estrategia final:")
print(f"   One-Hot Encoding: {columnas_one_hot}")
print(f"   Frequency Encoding: {columnas_frequency}")
print(f"   Eliminar: {columnas_eliminar}")
```

```python
# BLOQUE 5 - LIMPIEZA (OPTIMIZADO Y SEGURO)
from sklearn.impute import SimpleImputer

print("===  EMPEZANDO LA LIMPIEZA DE NULOS ===")

# ---------------------------------------------------------
# PASO CRÍTICO: ACTUALIZAR INVENTARIO DE NÚMEROS
# ---------------------------------------------------------
# ¿Por qué hacemos esto?
# Porque antes 'Value' y 'Wage' eran texto. Ahora que las convertimos a números,
# necesitamos que entren en la lista de "columnas_numericas" para poder
# calcular su mediana.
columnas_numericas = df.select_dtypes(include=['number']).columns.tolist()
print(f" Inventario actualizado: {len(columnas_numericas)} columnas numéricas detectadas.")


# ---------------------------------------------------------
# PASO 5.1: Rellenar NÚMEROS vacíos con la MEDIANA
# ---------------------------------------------------------
if columnas_numericas:
    # Usamos la mediana porque los salarios tienen outliers (Messi, CR7)
    # que dañarían el promedio.
    imputer = SimpleImputer(strategy='median')

    # Aplicamos la limpieza
    df[columnas_numericas] = imputer.fit_transform(df[columnas_numericas])
    print(f" Números vacíos rellenados con la mediana en {len(columnas_numericas)} columnas.")


# ---------------------------------------------------------
# PASO 5.2: Rellenar TEXTO vacío con "DESCONOCIDO"
# ---------------------------------------------------------
# Aquí usamos TUS listas de estrategia (que siguen intactas)
columnas_texto_a_limpiar = columnas_one_hot + columnas_frequency

if columnas_texto_a_limpiar:
    # Rellenamos solo las columnas de texto que vamos a usar
    df[columnas_texto_a_limpiar] = df[columnas_texto_a_limpiar].fillna('DESCONOCIDO')
    print(f" Texto vacío rellenado en {len(columnas_texto_a_limpiar)} columnas.")

print("\n ¡Limpieza de nulos completada!")
```

```python
# BLOQUE 6 - TRANSFORMACIONES INTELIGENTES (OPTIMIZADO)
# Nota: No escalamos aquí para mantener los datos legibles (Euros, Kg, CM) hasta el final.

print("===  TRANSFORMANDO LOS DATOS INTELIGENTEMENTE ===")

# PASO 6.1: One-Hot Encoding (Corregido)
if columnas_one_hot:
    print(f" Aplicando One-Hot Encoding a: {len(columnas_one_hot)} columnas...")
    # drop_first=True es OBLIGATORIO para evitar multicolinealidad
    df = pd.get_dummies(df, columns=columnas_one_hot, drop_first=True)

# PASO 6.2: Frequency Encoding (Corregido: Borra la original)
if columnas_frequency:
    print(f" Aplicando Frequency Encoding a: {len(columnas_frequency)} columnas...")
    for col in columnas_frequency:
        if col in df.columns:
            # 1. Calcular frecuencias
            frecuencias = df[col].value_counts(normalize=True)
            # 2. Mapear
            df[f'{col}_freq'] = df[col].map(frecuencias)
            # 3. BORRAR la original de texto (Crucial)
            df.drop(col, axis=1, inplace=True)
            # print(f"   Transformada: {col}") # Comentado para no llenar la pantalla

# PASO 6.3: Extraer info de fechas antes de borrar (Tu lógica avanzada)
# Esto intentará salvar info de columnas que vamos a eliminar
if columnas_eliminar:
    print(" Intentando rescatar fechas de columnas a eliminar...")
    for col in columnas_eliminar:
        if col in df.columns:
            # Detectar si parece fecha
            if any(x in col.lower() for x in ['date', 'fecha', 'joined', 'contract']):
                try:
                    # Intentamos convertir a datetime
                    fecha_temp = pd.to_datetime(df[col], errors='coerce')
                    # Si logramos extraer algo util (que no sean todo NaT)
                    if fecha_temp.notna().sum() > 0:
                        df[f'{col}_year'] = fecha_temp.dt.year.fillna(0).astype(int)
                        print(f"    ¡Rescatado Año de!: {col}")
                except:
                    pass # Si falla, no importa, igual se iba a eliminar

# PASO 6.4: Eliminar columnas problemáticas (Limpieza final)
columnas_a_eliminar_final = [col for col in columnas_eliminar if col in df.columns]

if columnas_a_eliminar_final:
    df = df.drop(columns=columnas_a_eliminar_final)
    print(f" Eliminadas {len(columnas_a_eliminar_final)} columnas de ruido (Nombres, Links, Texto viejo).")

print("\n" + "="*40)
print(f" PROCESO TERMINADO")
print(f"Filas: {df.shape[0]}, Columnas: {df.shape[1]}")

# Auditoría Final de Texto
cols_texto = df.select_dtypes(include=['object']).columns.tolist()
if len(cols_texto) == 0:
    print(" ESTADO: Dataset 100% Numérico. Listo para IA.")
else:
    print(f" ALERTA: Aún quedan columnas de texto: {cols_texto}")
```

```python
# BLOQUE DE PULIDO FINAL
# Objetivo: Eliminar identificadores y fechas crudas que ya procesamos

cols_a_sacar = ['ID', 'Joined']

print(f"Dimensiones antes: {df.shape}")

# Borramos solo si existen
cols_existentes = [c for c in cols_a_sacar if c in df.columns]
if cols_existentes:
    df.drop(cols_existentes, axis=1, inplace=True)
    print(f" Se eliminaron columnas sobrantes: {cols_existentes}")
else:
    print(" Ya no existían esas columnas.")

print(f"Dimensiones finales listas para IA: {df.shape}")
```
